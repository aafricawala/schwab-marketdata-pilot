```python
import os
import google.colab.userdata

# Set GitHub username
os.environ['GITHUB_USER'] = google.colab.userdata.get("GITHUB_USERNAME")

# Retrieve your GitHub PAT from Colab Secrets
os.environ['GITHUB_PAT'] = google.colab.userdata.get("GITHUB_TOKEN_SCH_MKT_DATA")

# Configure Git identity (required before committing)
!git config --global user.email "aafricawala@gmail.com"
!git config --global user.name "Ankit Africawala"

# Update remote URL to include PAT for non-interactive pushes
!git remote set-url origin https://$GITHUB_USER:$GITHUB_PAT@github.com/$GITHUB_USER/schwab-marketdata-pilot.git
```

```python
# Prompt for Schwab Client ID (stored only in memory)
os.environ["SCHWAB_CLIENT_ID"] = input("Enter Schwab Client ID: ")

# Prompt for Schwab Client Secret (stored only in memory)
os.environ["SCHWAB_CLIENT_SECRET"] = input("Enter Schwab Client Secret: ")

# Set Schwab Redirect URI (must match your Schwab Developer App)
os.environ["SCHWAB_REDIRECT_URI"] = "https://127.0.0.1"
```

```python
# Import the SchwabClient class from your project file.
# This file must exist in your GitHub repo and be pulled into Colab.
from schwab_client import SchwabClient

# Create a SchwabClient instance using your Client ID and Redirect URI.
# The Client ID identifies your app, and the Redirect URI must match Schwab's configuration.
client = SchwabClient(
    client_id=os.environ["SCHWAB_CLIENT_ID"],
    redirect_uri=os.environ["SCHWAB_REDIRECT_URI"]
)

# Build the OAuth authorization URL.
# This URL is where you will log in and approve access for your app.
auth_url = client.build_auth_url()

# Print the URL so you can click it or copy/paste it into your browser.
print(auth_url)
```

```python
# Prompt the user to paste the authorization code obtained from the browser redirect.
# This code is temporary and safe to paste here.
auth_code = input("Paste the authorization code from the redirect URL: ")

# Exchange the authorization code for an access token.
# The Client Secret is used here but never stored anywhere.
token_data = client.exchange_code_for_token(
    code=auth_code,
    client_secret=os.environ["SCHWAB_CLIENT_SECRET"]
)

# Print the token data so you can verify the access token was issued.
print(token_data)
```

```python
# Call the Schwab Market Data API to retrieve a quote for a specific symbol.
# The access token obtained earlier is automatically used by the client.
quote = client.get_quote("AAPL")

# Print the quote data returned by Schwab's API.
print(quote)
```
